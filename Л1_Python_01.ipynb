# 3: Циклы (for) — автоматизация повторений

## 1. Цикл for и функция range()

Аналогия: «Для каждого элемента в некоторой последовательности сделай вот это.»

```python
# Простейший цикл: распечатать числа от 1 до 5
for i in range(1, 6, 2):  # range(start, stop) – от start ДО stop (не включая stop!)
    print("Текущее число:", i)
```

```python
# Более математичный пример: таблица умножения на 7
print("Таблица умножения на 7:")
for multiplier in range(1, 11):
    result = 7 * multiplier
    print(f"7 x {multiplier} = {result}")
```

```python
# Цикл по готовой последовательности (списку)
list_of_squares = [1, 4, 9, 16, 25, 121]
for a in list_of_squares:
    print("Квадрат числа?", a)
```

## 2. Накопление результата в цикле (сумматор)

Ключевая концепция: Создаём переменную-«накопитель» (аккумулятор) до цикла и изменяем её внутри цикла.

```python
# Задача 1: Сумма всех чисел от 1 до N

N = int(input("Введите N "))
s = 0  # Начинаем с нуля (значение от чего оттталкиваемся)

for number in range(1, N + 1):
    s = s + number  # Или sum += number



print(f"Сумма чисел от 1 до {N} равна {s}")

# Сверяем с формулой Гаусса:
print("Проверка по формуле Гаусса:", N * (N + 1) // 2)
```

```python
# Задача 2: Поиск максимального элемента в списке (вручную)
numbers = [3, 7, 2, 9, 1, 4]
current_max = numbers[0]  # Предполагаем, что первый элемент — максимальный
for num in numbers:
    if num > current_max:
        current_max = num
print("Максимальное число в списке:", current_max)
print(len(numbers))
```

## 4.2 Методы строк и проверка условий

```python
raw_input = "   пИтОн ПрОгРаМмиРоВаНиЕ   "

# Очистка данных от лишних пробелов по краям
clean = raw_input.strip() 

# Работа с регистром
print(clean.lower())  # 'питон программирование'
print(clean.upper())  # 'ПИТОН ПРОГРАММИРОВАНИЕ'

# Поиск подстроки (самый правильный способ проверки — оператор in)
if "питон" in clean.lower():
    print("Слово найдено!")

# Замена символов. replace() возвращает НОВУЮ строку, исходная не меняется!
new_str = clean.replace(" ", "_")
print(new_str)        # 'пИтОн_ПрОгРаМмиРоВаНиЕ'

#Метод .join() проходит по списку один раз и сразу выделяет нужный объем памяти.
#Соединение без разделителя
letters = ['П', 'р', 'и', 'в', 'е', 'т'] 
word = "".join(letters) # Мы используем пустую строку "" как клей. Она ничего не вставляет между элементами. 
print(word) # Привет

#Соединение через пробел (самый частый случай)
words = ["Привет", "мир", "Python"] 
sentence = " ".join(words) # Пробел " " встает МЕЖДУ словами. 
print(sentence) # Привет мир Python

#Соединение через запятую (создание CSV-строки)
data = ["Иван", "Иванов", "150000"] 
csv_line = ",".join(data) 
print(csv_line) # Иван,Иванов,150000

# Проверки-предикаты
code = "A-777-БК"
print(code.isalnum())       # False (из-за дефиса)
print(any(c.isalpha() for c in code)) # True (есть ли хотя бы одна буква)
```
